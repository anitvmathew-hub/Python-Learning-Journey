## List (Creation, Modification and Access):

### 1. List Creation: 

```python
#a. Create a list named age_list with five integer elements.For eg., [24, 25, 26, 27, 28]
age_list=[24,25,26,27,28]

#b. Create a list named name_list with five string elements.
name_list=["Juan","Aarav","Ryan","Dane","Ezak"]

print("Age list:",age_list)
print("Name list:",name_list)
```

### 2. List Operations / Modifications:

```python
# a. Append "Yazhini" to name_list
name_list.append("Yazhini")
print("\nAfter appending 'Yazhini':", name_list)

# b. Insert 30 at index 2 in age_list
age_list.insert(2, 30)
print("After inserting 30 at index 2:", age_list)

# c. Remove "Yazhini" from name_list
name_list.remove("Yazhini")
print("After removing 'Yazhini':", name_list)

# d. Pop the last element from age_list
popped_value = age_list.pop()
print("After popping last element:", age_list)
print("Popped value:",popped_value)

# e. Extend age_list with [29, 30, 26]
age_list.extend([29, 30, 26])
print("After extending with [29, 30, 26]:", age_list)

# f. Sort age_list in descending order
age_list.sort(reverse=True)
print("After sorting in descending order:", age_list)

# g. Max, Min, and Sum of ages
max_age = max(age_list)
min_age = min(age_list)
sum_age = sum(age_list)
print("Max age:",max_age,"\nMin age:",min_age,"\nSum of ages:",sum_age)

```

### 3. Accessing List Elements

```python
#a.First element
print("First element:", name_list[0])

#b.Last element
print("Last element:", name_list[-1])

#c.Elements from index 2 to 4
print("Elements from index 2 to 4:", name_list[2:5])

#d.Elements in reverse order
print("Reversed name_list:", name_list[::-1])
```

## Dictionary (Creation, Modification and Access): 

```python
#a. Create a dictionary named student_marks that maps the names of fi ve students to their marks (use scale of from 0 to 100). 
student_marks = { "Juan": 93, "Aarav": 75, "Ryan": 83, "Dane": 68, "Ezak": 88}
print(student_marks)    

#b. Access and print the mark of a specific student, of your choice. 
print("Juan's mark:", student_marks["Juan"])   

#c. Add a new student "Janani" with a mark of 80 to the student_marks dictionary. 
student_marks["Janani"] = 80

#d. Update the mark of any one older student to 82.
student_marks["Dane"]=82
print("Dane's mark:",student_marks["Dane"])

#e. Use the keys(), values(), and items() methods to print all keys, values, and key-value pairs in the student_marks dictionary
print("Keys:", list(student_marks.keys()))
print("Values:", list(student_marks.values()))
print("Items:", list(student_marks.items()))
```

## Sets (Operations):

```python
#a. Create a set called my_set with following values:['a','e','i','o','u','a','a','i'].Analyse the output and provide explanation for the same.

my_set = set(['a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'])
print(my_set)
    

    
```

#### Explanation

##### In set , duplicate values are not allowed. In the given input list ['a','e','i','o','u','a','a','i'] there are duplicate values.A set can only store each distinct value once.So after removing duplicate values the set will be as {'a','i','e','o','u'}.


```python
#b. Attempt to change the value of my_set[4] = 's'. If code throws an error, provide an explanation.
my_set[4] = 's'    
```

#### Explanation:
##### Sets are unordered, meaning elements don't have a fixed position or index so there is no such thing as my_set[4].


```python
#c. Create two sets:
#set1 with values: {1, 3, 5, 7, 9}
#set2 with values: {2, 3, 5, 8, 10}

set1={1, 3, 5, 7, 9}
set2={2, 3, 5, 8, 10}

```

```python
#d. Compute and print the union and intersection of set1 and set2.

union = set1.union(set2)
intersection = set1.intersection(set2)

print("Result of union of set 1 and set 2:", union)
print("Result of intersection of sert 1 and set 2:", intersection)
```

## Operators & Conditional Statements :(IF, ELIF, ELSE)

### Performance Category Program:

```python
score = int(input("Enter your score (0 to 10): "))

if 0 <= score <= 10:
    if score > 7:
        print("Above Average: Excellent work!")
    elif 4 <= score <= 7:
        print("Average: Good effort! Keep practicing, there is room for improvement.")
    else:
        print("Below Average: Need to improve your performance, consistent practice will lead to better results.")
else:
    print("Invalid input! Please enter a score between 0 and 10.")
```
