## Numpy Array Operations:

```python
import numpy as np
import pandas as pd

# 1. Create a 1D NumPy Array:
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("Temperatures for week 1:", temperatures_w1)
experience=[1,2,3,4,5]
salary=[30000,35000,40000,48000,55000]

# 2. Inspection and Properties:
print("\nShape of the array:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)

# 3. Array Operations:
# Fahrenheit = (Celsius * 9/5) + 32
fahrenheit = (temperatures_w1 * 9 / 5) + 32
print("\nFahrenheit:", fahrenheit)
print("Max:", temperatures_w1.max())
print("Min:", temperatures_w1.min())
mean=temperatures_w1.mean()
print("Mean:", round(mean,2))

# 4. Array Slicing and Indexing:
print("\n\nTemperatures for the first three days of the week:",temperatures_w1[:3])
print("Temperatures for the weekend (last two days):",temperatures_w1[-2:])
print("Temperatures for the middle three days of the week:",temperatures_w1[2:5])

# 5. Create a 2D Array:
temperatures = np.array([ [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
                          [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] ])

# 6. Inspect and Slice the 2D Array:
print("\n\nShape of the temperatures array:",temperatures.shape)
print("Data Type of the temperatures array:",temperatures.dtype)
print("Total number of elements in the temperatures array:",temperatures.size)
print("Temperature of Week 1:", temperatures[0])
print("Temperature of Week 2:", temperatures[1])
print("\n\nWeekend  Temperature of week 1 :", temperatures[0, -2:])
print("Weekend  Temperature of week 2 :", temperatures[1, -2:])

```

## Pandas Series:

```python
# 1. Creating Pandas Series:
marks=pd.Series([95, 92, 89, 85, 80],index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print("\nList of Ranks & Marks\n")
print(marks)

#2. Indexing and Slicing:
print("\nMark of the first rank (using integer index position):", marks.iloc[0])
print("\nMarks of the top 3 ranks by specifying their index labels:")
print(marks.loc['Rank1':'Rank3'])

print("\nMark of the 3rd rank student(using iloc):",marks.iloc[2])

print("\nThe list of ranks where the marks are greater than 90:\n")
print([i for i in marks.index if marks[i] > 90])   

#3. Manipulating Series:
marks['Rank1']=100
print("\nMark of the student with the first rank after modifying mark:",marks['Rank1'])
print("\nMark of the students after modifying mark of Rank1:")
print(marks)

#Remove the entry corresponding to the last rank from the Series
marks = marks.drop('Rank5')
print("\nAfter removing last rank:\n", marks)

cgpa = marks / 10
print("\nRank     CGPA")
print("**************")
print(cgpa)

```

## Pandas DataFrame:

```python
#1. Creating Pandas DataFrame:
dict={'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
      'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture','Clothing', 'Electronics', 'Furniture', 'Clothing','Furniture', 'Electronics'],
      'Region': ['North', 'South', 'North', 'East', 'West','North', 'East', 'West', 'South', 'North'],
      'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]}
transactions = pd.DataFrame(dict)

# 2. Data Exploration:

print("\nDataFrame:\n", transactions)
print("\nHead:\n", transactions.head())
print("\nTail:\n", transactions.tail())
print("\nShape:", transactions.shape)
print("\nColumns:", list(transactions.columns))
print("\nData types:\n", transactions.dtypes)
transactions.info()

print("\n\nProductCategory and Amount:\n", transactions[['ProductCategory', 'Amount']])
print("\n\nLast 3 columns (iloc):\n", transactions.iloc[:, -3:])
print("\n\nNorth region and Amount > 200:\n",
      transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)])
print("\n\nValue counts of ProductCategory:\n", transactions['ProductCategory'].value_counts())
print("\n\nUnique regions:", transactions['Region'].unique())
print("\n\nMean amount by region:\n", transactions.groupby('Region')['Amount'].mean())


#3. Manipulating the DataFrame:

transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print("\n\nAfter updation \n", transactions)

transactions['Discount'] = transactions['Amount'] * 0.10
print("\n\nAfter adding Discount column:\n", transactions)

transactions = transactions[transactions['TransactionID'] != 109]
print("\n\nAfter removing TransactionID 109:\n", transactions)

transactions = transactions.drop(columns='Discount')
print("\n\nAfter deleting Discount column:\n", transactions)
```
